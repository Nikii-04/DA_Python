```python
# ASSIGN DATETIME DATATYPE TO job_posted_date(ITS A STR ORIGINALLY) DIRECTLY IN DATASET
# object turned into datetime

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
#month is not included in dataset to add new col in dataset create job_posted_month new col

df['job_posted_month'] = df.job_posted_date.dt.month
```

```python
#to sort in dataset
# use -> df = df.sort_values("job_posted_date") there is another method #inplace

df.sort_values(by='job_posted_date', inplace=True)
```

```python
# droppping col -> axis= 0 or index, 1 or columns, default 0
df.drop(labels='salary_hour_avg', axis=1)
```

```python
# still col is there
df.info()
```

```python
# to remove directly from dataset use inplace
df.drop(labels='salary_hour_avg', axis=1, inplace=True)
```

```python
df.info()
```

```python
# remove non values/ missing val i.e. NaN (drop complete row with NaN val for a specific col)
# subset is a list of cols
df.dropna(subset=['salary_year_avg'])
```

```python
df.info()
```

```python
df.dropna(subset=['salary_year_avg'], inplace=True)
```
