```python
from datasets import load_dataset
```

```python
dataset = load_dataset('lukebarousse/data_jobs')
```

```python
import pandas as pd
```

```python
df = dataset['train'].to_pandas()
```

```python
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
job_counts = df.job_title_short.value_counts()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.bar(job_counts.index, job_counts)
plt.title('posting by job titles')
plt.ylabel('counts of job postings')
plt.show()
```

```python
plt.bar(job_counts.index, job_counts)
plt.title('posting by job titles')
plt.ylabel('counts of job postings')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(job_counts.index, job_counts)
plt.title("postings by job title")
plt.ylabel('counts of postings')
plt.xticks(rotation=45, ha='right')   #ha=horizontal aignment - align name with graph properly
```

# Matplotlib pandas ***plotting***

# ***Median salary vs data science jobs***

```python
df.groupby('job_title_short')['salary_year_avg'].median()
```

```python
df.groupby('job_title_short')['salary_year_avg'].median().sort_values()
```

```python
df.groupby('job_title_short')['salary_year_avg'].mean().sort_values()
```

```python
job_salary = df.groupby('job_title_short')['salary_year_avg'].median().sort_values()
```

```python
job_salary.plot(kind='barh')
plt.xlabel('salary ($USD) ')
plt.ylabel('job_title')
plt.title('median salary by job title')
plt.show()
```
