```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from datasets import load_dataset

dataset = load_dataset("lukebarousse/data_jobs")
df = dataset['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_US = df[df['job_country'] == 'United States'].copy()
```

```python
df_US
```

Extract /covert job month no into full month name  '%B'

```python
print(df_US['job_posted_date'].dtype)
```

```python
print(df_US['job_posted_date'].head())
```

```python
print(df['job_posted_date'].dtype)
print(df['job_posted_date'].head())
```

```python
df_US['job_posted_month'] = df_US['job_posted_date'].dt.strftime('%B')
```

```python
df_US
```

count of job posting by time/month

```python
df_US_pivot = df_US.pivot_table(index='job_posted_month', columns='job_title_short', aggfunc='size')
df_US_pivot
```

```python
df_US_pivot.reset_index(inplace=True)
```

```python
df_US_pivot
```

```python
df_US_pivot['month_no'] = pd.to_datetime(df_US_pivot['job_posted_month'], format='%B')
df_US_pivot['month_no']
```

extract month no

```python
df_US_pivot['month_no'] = pd.to_datetime(df_US_pivot['job_posted_month'], format='%B').dt.month
```

```python
df_US_pivot
```

```python
df_US_pivot.sort_values('month_no', inplace=True)
```

```python
df_US_pivot
```

```python
df_US_pivot.set_index('job_posted_month', inplace=True)
```

```python
df_US_pivot
```

```python
df_US_pivot.drop(columns='month_no', inplace=True)
```

```python
df_US_pivot
```

```python
df_US_pivot.plot(kind='line')
```
